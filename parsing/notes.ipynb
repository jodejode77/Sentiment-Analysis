```python
import pandas as pd

# Load the CSV file
file_path = "test.csv"  # Update with your file path
df = pd.read_csv(file_path)

# Add a new column for sentiment
df['sentiment'] = None

# Track how many texts have been printed
text_count = 0

# Iterate over the rows and print the text for sentiment analysis
for index, row in df.iterrows():
    text_count += 1
    formatted_text = row['text'].replace('.', '.\n')
    print(f"Text #{text_count}:\n{formatted_text}")
    
    # Input sentiment with validation
    while True:
        sentiment = input("Enter sentiment (1 for positive, 0 for neutral, -1 for negative, or 'finish' to stop): ")
        
        if sentiment.lower() == 'finish':
            # If user types 'finish', exit the loop
            break
        elif sentiment in ['1', '0', '-1']:
            # If input is valid, store it
            df.at[index, 'sentiment'] = int(sentiment)
            break
        else:
            # Invalid input, prompt again
            print("Invalid input. Please enter 1, 0, or -1 for sentiment or 'finish' to stop.")

    if sentiment.lower() == 'finish':
        # If 'finish' is entered, break out of the main loop
        break

# Save the updated DataFrame to a new CSV file
df.to_csv("test.csv", index=False)

# Print the number of texts processed
print(f"\nNumber of texts printed: {text_count}")
```

```python
import pandas as pd
import glob
import re

def remove_duplicates_from_csv():
    # Define the regex pattern for file matching
    pattern = re.compile(r'data_(AFLT|SBER|MTSS|LKOH|MOEX|GMKN)_train\.csv')
    
    # Get all matching CSV files in the directory
    files = [f for f in glob.glob("data_*_train.csv") if pattern.match(f)]
    
    for file in files:
        # Read the CSV file
        df = pd.read_csv(file)
        
        # Count duplicates before removing them
        initial_count = len(df)
        df = df.drop_duplicates(subset=['text'], keep='first')
        final_count = len(df)
        deleted_count = initial_count - final_count
        
        # Save back to the same filename
        df.to_csv(file, index=False)
        print(f"Processed and saved: {file} | Duplicates removed: {deleted_count}")

remove_duplicates_from_csv()
```

```python
import pandas as pd
import re
import spacy

def clean_text(text, ticker):
    """
    Removes unnecessary symbols like {$TICKER} and applies context windowing.
    """
    nlp = spacy.load("ru_core_news_lg")
    
    text = re.sub(r"\{\$(.*?)\}", r"\1", text)  # Converts {$TICKER} → TICKER
    
    # Split text into sentences
    doc = nlp(text)
    sentences = [sent.text for sent in doc.sents]
    
    # Find sentences mentioning the given ticker
    relevant_indices = [i for i, s in enumerate(sentences) if ticker in s]
    
    # Extract context window (2 sentences before and 2 after)
    extracted_sentences = []
    for idx in relevant_indices:
        start = max(0, idx - 2)
        end = min(len(sentences), idx + 3)
        extracted_sentences.extend(sentences[start:end])
    
    return " ".join(extracted_sentences)

def process_csv(file_path, ticker):
    """
    Reads the CSV file, cleans the text, applies context windowing, and saves it back.
    """
    df = pd.read_csv(file_path)
    
    df["text"] = df["text"].apply(lambda x: clean_text(str(x), ticker))
    
    df.to_csv("testMTSS.csv", index=False)
    print(f"Processed and saved: testMTSS.csv")

process_csv("data_MTSS_train.csv", "MTSS")
```
